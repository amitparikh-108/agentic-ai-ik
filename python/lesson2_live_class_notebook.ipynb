````python
def route_summary(route_id, show_stops=True, symbol="₹"):
    """Return a formatted one-line summary of a route."""
    if route_id not in NETWORK:
        return f"Route {route_id} not found"
    info = NETWORK[route_id]
    ac   = "AC" if info["is_ac"] else "Non-AC"
    line = f"Route {route_id} | {info['name']} | {ac} | {symbol}{info['base_fare']}"
    if show_stops:
        line += f" | Stops: {" → ".join(info['stops'])}"
    return line

print("=== Default args ===")
print(route_summary("12"))                          # show_stops defaults to True
print(route_summary("12", show_stops=False))        # hide stops
print(route_summary("12", show_stops=False, symbol="$"))  # USD
````

### --- 2. *args: compare any number of routes ---


````python
def compare_routes(*route_ids, metric="base_fare"):
    """Compare any number of routes on a given metric."""
    print(f"\n=== Comparing routes by: {metric} ===")
    results = []
    for rid in route_ids:
        if rid in NETWORK:
            val = NETWORK[rid].get(metric, NETWORK[rid]["base_fare"])
            results.append((rid, NETWORK[rid]["name"], val))
    for i, (rid, name, val) in enumerate(sorted(results, key=lambda x: -x[2]), 1):
        print(f"  {i}. Route {rid} ({name}): {val}")

compare_routes("7", "12", "31", "88")
compare_routes("7", "12", metric="capacity")
````

````python
def filter_trips(**criteria):
    """
    Return trips matching ALL provided criteria.
    Accepted keys: route_id (str), is_peak (bool), is_ac (bool), min_passengers (int)
    """
    results = TRIP_LOG.copy()
    if "route_id" in criteria:
        results = [t for t in results if t["route_id"] == criteria["route_id"]]
    if "is_peak" in criteria:
        results = [t for t in results if t["is_peak"] == criteria["is_peak"]]
    if "is_ac" in criteria:
        results = [t for t in results if t["is_ac"] == criteria["is_ac"]]
    if "min_passengers" in criteria:
        results = [t for t in results if t["passengers"] >= criteria["min_passengers"]]
    return results

print("\n=== **kwargs filtering ===")
peak_ac = filter_trips(is_peak=True, is_ac=True)
print(f"Peak + AC trips    : {len(peak_ac)}")
busy_31 = filter_trips(route_id="31", min_passengers=70)
print(f"Route 31, 70+ pax  : {len(busy_31)}")
````

| Concept | What it is | Where you used it |

| `list` | Ordered, mutable sequence | Stop lists, TRIP_LOG, result sets |

| List comprehension | One-line transform/filter | AC routes, fare lists, load alerts |

| `.extend()` | Add all items of another list | Building all_stops |
| `dict` | Key→value lookup | NETWORK, revenue_by_route, hourly counts |
| `.get(key, default)` | Safe dict access | Accumulator pattern |
| `.items()` | Iterate key-value pairs | All analytics loops |
| Dict comprehension | Build dicts in one line | name_to_id, ac_fares |
| `set` | Unique values, no order | Unique stops, operator coverage |
| `\|` `&` `-` `^` | Set operations | Venn-diagram stop analysis |
| `.update()` | Add all items of iterable to set | Building stop sets |
| `def` | Define a named function | All analytics functions |
| Default args `def f(x, n=3)` | Optional parameters | top_n, metric, symbol |
| `*args` | Variable positional args as tuple | `compare_routes(*route_ids)` |
| `**kwargs` | Variable keyword args as dict | `revenue_breakdown(**filters)` |
| `return` | Send a value back to the caller | Every analytics function |
| Scope (LEGB) | Where variables are visible | Global TRIP_LOG, local accumulators |
| `lambda` | Anonymous one-line function | `key=` in sort, filter, map |
